```python
import os
import pandas as pd

RAW_DIR = "../data/raw"
files = {
    'membres': 'membres.xlsx',
    'contrats': 'contrats.xlsx',
    'contrat_membres': 'contrat_membres.xlsx',
    'cotisations': 'cotisations.xlsx',
    'recettes': 'recettes.xlsx',
    'sinistres_adm': 'sinistres_adm.xlsx',
    'sinistres_recu': 'sinistres_recu.xlsx'
}

audit_summary = []

for name, filename in files.items():
    filepath = os.path.join(RAW_DIR, filename)
    if os.path.exists(filepath):
        xls = pd.ExcelFile(filepath)
        sheet_name = xls.sheet_names[0]
        df = pd.read_excel(filepath, sheet_name=sheet_name)
        
        audit_summary.append({
            'Fichier': filename,
            'Feuille': sheet_name,
            'Lignes': df.shape[0],
            'Colonnes': df.shape[1],
            'Doublons Stricts': df.duplicated().sum(),
            'Cellules Manquantes': df.isnull().sum().sum()
        })

df_audit = pd.DataFrame(audit_summary)
print("=== RAPPORT D'AUDIT GLOBAL ===")
display(df_audit)
```
